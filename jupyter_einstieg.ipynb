# Jupyter kennenlernen: Temperaturen auswerten

Jupyter-Notebooks verbinden **Text, Python-Code und Ergebnisse** in einem Dokument. Du kannst damit Ideen ausprobieren, Daten untersuchen und deine Schritte verständlich dokumentieren. Besonders praktisch ist das für Datenanalyse, Unterricht, Forschung und Prototypen.

In diesem Beispiel sehen wir uns die Temperaturen einer Woche an. Führe die Codezellen der Reihe nach aus: Klicke in eine Zelle und drücke **Shift+Enter**.

## 1. Daten eingeben

Eine Python-Liste ist hier unser kleiner Beispieldatensatz. Die Werte stehen für die Höchsttemperatur an sieben aufeinanderfolgenden Tagen, in Grad Celsius.

```python
temperaturen = [18, 21, 19, 23, 20, 17, 22]
tage = ["Montag", "Dienstag", "Mittwoch", "Donnerstag", "Freitag", "Samstag", "Sonntag"]

for tag, temperatur in zip(tage, temperaturen):
    print(f"{tag}: {temperatur} °C")
```

## 2. Fragen an die Daten stellen

In einem Notebook kannst du kleine Berechnungen einzeln ausprobieren. Ändere zum Beispiel später einen Temperaturwert und führe diese Zellen erneut aus.

```python
kaeltester_tag = tage[temperaturen.index(min(temperaturen))]
waermster_tag = tage[temperaturen.index(max(temperaturen))]
durchschnitt = sum(temperaturen) / len(temperaturen)

print(f"Kältester Tag: {kaeltester_tag} ({min(temperaturen)} °C)")
print(f"Wärmster Tag: {waermster_tag} ({max(temperaturen)} °C)")
print(f"Durchschnitt: {durchschnitt:.1f} °C")
```

## 3. Werte sichtbar machen

Auch eine einfache Textgrafik kann ein Muster zeigen. Jeder Balken steht für einen Tag; ein Zeichen entspricht ungefähr einem Grad.

```python
for tag, temperatur in zip(tage, temperaturen):
    balken = "#" * temperatur
    print(f"{tag:<10} {balken} {temperatur} °C")
```

## Was du über Jupyter wissen solltest

- **Zelle:** Ein Notebook besteht aus einzelnen Bereichen. Markdown-Zellen enthalten Erklärungen; Codezellen enthalten ausführbaren Code.
- **Kernel:** Das ist der laufende Python-Prozess, der deinen Code ausführt und Variablen wie `temperaturen` im Arbeitsspeicher behält.
- **Ausführen:** Mit **Shift+Enter** wird eine Zelle ausgeführt und zur nächsten gewechselt. Das Ergebnis erscheint direkt darunter.
- **Reihenfolge:** Zellen können einzeln und in beliebiger Reihenfolge gestartet werden. Der Kernel behält Änderungen aber in der tatsächlich ausgeführten Reihenfolge. Wenn Ergebnisse verwirrend werden, starte den Kernel neu und führe die Zellen von oben nach unten aus.
- **Speichern:** Speichere das Notebook wie eine normale Datei. Der Code und die Erklärungen bleiben erhalten; Ausgaben können ebenfalls mitgespeichert werden.

**Typische Einsatzzwecke:** Daten erkunden und visualisieren, Berechnungen dokumentieren, Code im Unterricht erklären und kleine Ideen testen.
